```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Bibliotecas carregadas!")
```

```python

dados = pd.read_csv(
    "Balanco-2019-Ligue180-Tipo de Violação (1).csv",
    encoding="utf-8-sig",
    header=9,
    thousands=",",
    na_values=["N/A"]
)

# Remover coluna vazia, se existir
dados = dados.loc[
    :, ~dados.columns.astype(str).str.startswith("Unnamed")
]

display(dados.head(10))

print("Tamanho da tabela:", dados.shape)
print("Colunas:", dados.columns.tolist())

```

```python

print("=== TAMANHO DA BASE ===")
print("Linhas e colunas:", dados.shape)

print("\n=== NOMES DAS COLUNAS ===")
print(dados.columns.tolist())

print("\n=== VALORES VAZIOS ===")
print(dados.isnull().sum())

print("\n=== LINHAS DUPLICADAS ===")
print("Quantidade:", dados.duplicated().sum())

print("\n=== TIPOS DE DADOS ===")
print(dados.dtypes)

print("\n=== RESUMO DA BASE ===")
dados.info()
```
